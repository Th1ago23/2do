```
import pandas as pd

# Carregando os arquivos
sap_df = pd.read_csv("Sapnaoprojecao.csv")
proj_df = pd.read_csv("Projecaonaosap.csv")

# Renomeando e padronizando
def preparar_df(df, col_rename):
    df = df.rename(columns=col_rename)
    for col in ['Valor_Projecao', 'Valor_SAP']:
        df[col] = df[col].fillna("0").astype(str).str.replace("R\$", "", regex=True).str.replace(".", "", regex=False).str.replace(",", ".", regex=False)
        df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0)
    return df

sap_df = preparar_df(sap_df, {
    'Elemento PEP': 'Elemento_PEP',
    'Nº doc.de referência': 'Doc_Referencia',
    'Soma de Valor Plan Linha': 'Valor_Projecao',
    'Soma de Valor Comp. SAP': 'Valor_SAP'
})

proj_df = preparar_df(proj_df, {
    'Elemento PEP': 'Elemento_PEP',
    'Nº doc.de referência': 'Doc_Referencia',
    'Soma de Saldo projeção': 'Valor_Projecao',
    'Soma de Valor Comp SAP': 'Valor_SAP'
})

# Fazendo merge para encontrar os comuns
merged = pd.merge(proj_df, sap_df, on=["Elemento_PEP", "Doc_Referencia"], how="outer", suffixes=('_proj', '_sap'))

# Corrigindo os valores SAP
merged["Valor_Projecao_Final"] = merged["Valor_Projecao_proj"].fillna(merged["Valor_Projecao_sap"])
merged["Valor_SAP_Final"] = merged["Valor_SAP_proj"].fillna(0) + merged["Valor_SAP_sap"].fillna(0) - merged["Valor_SAP_proj"].fillna(0)

# Selecionando o resultado final
resultado = merged[["Elemento_PEP", "Doc_Referencia", "Valor_Projecao_Final", "Valor_SAP_Final"]]

# Exportando
resultado.to_csv("Resultado_Consolidado_Corrigido.csv", index=False, encoding="utf-8-sig")
```

```
import pandas as pd

# Carregando os arquivos
sap_df = pd.read_csv("Sapnaoprojecao.csv")
proj_df = pd.read_csv("Projecaonaosap.csv")

# Padronizando os DataFrames
sap_df = preparar_df(sap_df, {
    'Elemento PEP': 'Elemento_PEP',
    'Nº doc.de referência': 'Doc_Referencia',
    'Soma de Valor Plan Linha': 'Valor_Projecao',
    'Soma de Valor Comp. SAP': 'Valor_SAP'
})

proj_df = preparar_df(proj_df, {
    'Elemento PEP': 'Elemento_PEP',
    'Nº doc.de referência': 'Doc_Referencia',
    'Soma de Saldo projeção': 'Valor_Projecao',
    'Soma de Valor Comp SAP': 'Valor_SAP'
})

# Renomeando e padronizando
def preparar_df(df, col_rename):
    df = df.rename(columns=col_rename)
    for col in ['Valor_Projecao', 'Valor_SAP']:
        df[col] = df[col].fillna("0").astype(str)
        df[col] = df[col].str.replace("R\$", "", regex=True).str.replace(".", "", regex=False).str.replace(",", ".", regex=False)
        df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0)
    return df



# Fazendo merge
merged = pd.merge(proj_df, sap_df, on=["Elemento_PEP", "Doc_Referencia"], how="outer", suffixes=('_proj', '_sap'))

# Calculando os valores finais
merged["Valor_Projecao_Final"] = merged["Valor_Projecao_proj"].fillna(merged["Valor_Projecao_sap"])
merged["Valor_SAP_Final"] = merged["Valor_SAP_proj"].fillna(0) + merged["Valor_SAP_sap"].fillna(0) - merged["Valor_SAP_proj"].fillna(0)

# Selecionando e arredondando
resultado = merged[["Elemento_PEP", "Doc_Referencia", "Valor_Projecao_Final", "Valor_SAP_Final"]]
resultado["Valor_Projecao_Final"] = resultado["Valor_Projecao_Final"].round(2)
resultado["Valor_SAP_Final"] = resultado["Valor_SAP_Final"].round(2)

# Exportando direto pro Excel
resultado.to_excel("Resultado_Consolidado_Corrigido.xlsx", index=False, sheet_name="Consolidado")

print("Arquivo Excel criado com sucesso! 🎉")
```
